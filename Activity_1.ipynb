```python
#import the pandas libary
import pandas as pd
#load the dataset you created
df = pd.read_csv('campus_energy1.csv')
#Display the first few rows of the data
print("Here is the raw data:")
display(df.head())
#group the data to see total usage of the data
building_total = df.groupby('Building')['KWH_USE'].sum()
print("Total energy used by each building:")
print(building_total)
print("\nStatistical Summary:")
display(df.describe())
```
